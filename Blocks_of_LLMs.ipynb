```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")
text = "HI, Whatsup!"
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
import torch
from transformers import AutoModel

model = AutoModel.from_pretrained("gpt2")
inputs = tokenizer("HI, whatsup!", return_tensors="pt")
outputs = model(**inputs)
embeddings = outputs.last_hidden_state
print(embeddings.shape)  
```

```python
from transformers import AutoModelForCausalLM

model = AutoModelForCausalLM.from_pretrained("gpt2")
inputs = tokenizer("I dont give a", return_tensors="pt")
outputs = model.generate(**inputs, max_length=10)
print(tokenizer.decode(outputs[0]))
```

```python
import torch
import torch.nn as nn

class TinyLLM(nn.Module):
    def __init__(self, vocab_size, embed_size, num_heads, num_layers):
        super(TinyLLM, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_size)
        self.transformer = nn.Transformer(
            d_model=embed_size,
            nhead=num_heads,
            num_encoder_layers=num_layers,
            num_decoder_layers=num_layers,
        )
        self.fc = nn.Linear(embed_size, vocab_size)

    def forward(self, src, tgt):
        src = self.embedding(src)
        tgt = self.embedding(tgt)
        output = self.transformer(src, tgt)
        return self.fc(output)
    
vocab_size = 10000
embed_size = 512    
num_heads = 8       
num_layers = 6      

model = TinyLLM(vocab_size, embed_size, num_heads, num_layers)
```

```python

src = torch.randint(0, vocab_size, (10, 32))
tgt = torch.randint(0, vocab_size, (10, 32))
output = model(src, tgt)
print(output.shape)
```

```python
def generate_text(model, start_text, max_length=50):
    model.eval()
    tokens = tokenizer.encode(start_text)
    for _ in range(max_length):
        with torch.no_grad():
            output = model(torch.tensor([tokens]), torch.tensor([tokens]))
        next_token = output.argmax(dim=-1)[:, -1].item()
        tokens.append(next_token)
    return tokenizer.decode(tokens)

print(generate_text(model, "Once upon a time"))
```
